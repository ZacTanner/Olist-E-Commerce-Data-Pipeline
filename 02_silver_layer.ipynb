# Olist E-Commerce: Silver Layer
**Dataset**: [Brazilian E-commerce Public Dataset by Olist](https://www.kaggle.com/datasets/olistbr/brazilian-ecommerce)     
**Layer**: Silver: Cleaned, validated, and enriched  
**Tools Used**: Databricks, PySpark, Delta Lake  

## Overview
This notebook transforms the raw bronze tables into cleaned, validated, and analytically ready silver tables. Each table is processed independently through the following workflow: load from bronze, inspect, validate, transform, and write to silver as a Delta table.

The silver layer resolves data quality issues present in the Olist source data, including inconsistent/erroneous city name formatting, mixed-type fields, invalid coordinates, nonsensical review scores, and incomplete payment records.

## Processing Workflow (Per Table)
1. Load bronze table
2. Inspect schema and preview data
3. Check for NULL values
4. Validate primary and foreign keys
5. Apply cleaning and transformation logic
6. Add audit columns
7. Write to silver Delta table

## Tables Processed
- [`silver_customers`](#customers) — Geographic standardisation and region mapping
- [`silver_geolocation`](#geolocation) — Coordinate validation, deduplication by ZIP average
- [`silver_order_items`](#order-items) — Price validation, total item value derived
- [`silver_order_payments`](#order-payments) — Payment type standardisation, invalid payment removal
- [`silver_order_reviews`](#order-reviews) — Review score cleaning, malformed row removal
- [`silver_orders`](#orders) — Status consolidation, date extraction, delivery performance flag
- [`silver_product_category_translation`](#product-category-translation) — Reference table, no changes
- [`silver_products`](#products) — Category translation and consolidation, dimension validation, weight conversion
- [`silver_sellers`](#sellers) — Geographic standardisation and region mapping

## Cross-Table Validations
Foreign key integrity is validated across all 6 critical relationships after all tables are written. Payment totals are reconciled against order item totals with a 1-cent rounding tolerance.

## Shared Definitions and Functions

Utility objects used across multiple tables. Defined once here and referenced throughout the notebook to avoid repetition.

- **`valid_states`**: List of all 27 official Brazilian state codes used to validate customer, seller, and geolocation state fields.
- **`region_mapping`**: Maps each state code to one of Brazil's five macro-regions (North, Northeast, Central-West, Southeast, South).
- **`standardize_city_name()`**: UDF that normalises Brazilian city names by removing accents, apostrophes, and hyphens, collapsing whitespace, and applying title case. Applied to customers, geolocation, and sellers.
- **`city_map`**: Dictionary of high-confidence municipality name corrections derived from known data quality issues in the Olist dataset.

```python
valid_states = ["AC", "AL", "AP", "AM", "BA", "CE", "DF", "ES", "GO", "MA", 
                "MT", "MS", "MG", "PA", "PB", "PR", "PE", "PI", "RJ", "RN", 
                "RS", "RO", "RR", "SC", "SP", "SE", "TO"]
```

```python
region_mapping = {
    "AC": "North", "AM": "North", "AP": "North", "PA": "North", "RO": "North", 
    "RR": "North", "TO": "North",
    "AL": "Northeast", "BA": "Northeast", "CE": "Northeast", "MA": "Northeast",
    "PB": "Northeast", "PE": "Northeast", "PI": "Northeast", "RN": "Northeast", "SE": "Northeast",
    "DF": "Central-West", "GO": "Central-West", "MT": "Central-West", "MS": "Central-West",
    "ES": "Southeast", "MG": "Southeast", "RJ": "Southeast", "SP": "Southeast",
    "PR": "South", "RS": "South", "SC": "South"
}
```

```python
#City standardization function
def standardize_city_name(city_name):
    """
    Standardizes Brazilian city names by:
    1. Removing accents
    2. Removing apostrophes
    3. Replacing hyphens with spaces
    4. Normalizing whitespace
    5. Converting to title case
    """
    if city_name is None:
        return None

    # Remove accents
    city = unicodedata.normalize('NFKD', city_name)
    city = ''.join(
        c for c in city
        if not unicodedata.combining(c)
    )

    # Remove apostrophes / accent apostrophes
    city = (
        city
        .replace("'", "")
        .replace("´", "")
        .replace("`", "")
    )

    # Replace hyphens with spaces
    city = city.replace("-", " ")

    # Normalize "d ajuda" variations
    city = re.sub(
        r"\bd\s*ajuda\b",
        "d ajuda",
        city,
        flags=re.IGNORECASE
    )

    # Normalize D Oeste / Doeste
    city = re.sub(
        r"\bd\s*oeste\b",
        "doeste",
        city,
        flags=re.IGNORECASE
    )

    # Remove extra whitespace
    city = re.sub(r"\s+", " ", city)
    city = city.strip()

    # Convert to title case
    city = city.title()

    return city


# Register UDF
standardize_city_udf = udf(
    standardize_city_name,
    StringType()
)

# City consolidations
city_map = {
    # Shorthand fixes
    "Santa Barbara Doeste": "Santa Barbara Doeste",
    "Santa Barbara D Oeste": "Santa Barbara Doeste",
    "Embu": "Embu Das Artes",
    "Parati": "Paraty",
    "Piumhii": "Piumhi",
    "Bonfim Paulista": "Ribeirao Preto",
    "Holambra Ii": "Holambra",
    "Sao Bernardo": "Sao Bernardo Do Campo",
    "Jacare": "Jacarei",
    "Ribeirao": "Ribeirao Preto",
    "Santa Barbara": "Santa Barbara Doeste",
    "Valparaiso": "Valparaiso De Goias",
    "Paraguacu": "Paraguacu Paulista",
    # District/locality -> municipality
    "Goitacazes": "Campos Dos Goytacazes",
    "Papucaia": "Cachoeiras De Macacu",
    "Japuiba": "Angra Dos Reis",
    "Ipiabas": "Barra Do Pirai",
    "Conservatoria": "Valenca",
    "Itaipava": "Petropolis",
    "Penedo": "Itatiaia",
    "Barra De Sao Joao": "Casimiro De Abreu",
    "Vila Muriqui": "Mangaratiba",
    # Tourist districts
    "Arraial D Ajuda": "Porto Seguro",
    "Trancoso": "Porto Seguro",
    # Infrastructure / regional entities
    "Hidreletrica Tucurui": "Tucurui",
    "Carajas": "Parauapebas"
}

# Mapping expression
mapping_expr = F.create_map(
    [F.lit(x) for x in chain(*city_map.items())]
)

# Test Case
test_cases = [
    "arraial d ajuda",
    "arraial d'ajuda",
    "arraial d´ajuda",
    "arraial d`ajuda",
    "sao paulo",
    "SAO PAULO  ",
    "barra d oeste",
    "  rio de janeiro  ",
    "mogi-guacu"
]

print("City Standardization Test Results:")
print("=" * 60)

for test in test_cases:
    standardized = standardize_city_name(test)
    print(f"{test!r:30} -> {standardized!r}")

print("\n✓ Standardization function and mappings ready")
```

### Customers

**Source**: `bronze_customers` → `silver_customers`  
**Primary key**: `customer_id` (unique per transaction address)  
**Note on `customer_unique_id`**: Identifies the same physical person across 
multiple `customer_id` values. Use `customer_unique_id` for customer-level 
lifetime analysis; use `customer_id` as the join key to the orders table.

**Transformations applied**:
- City names standardised using `standardize_city_name()` UDF and `city_map` 
  consolidation dictionary
- Zip codes validated against Brazilian range (1000–99999)
- State codes validated against all 27 official Brazilian state codes
- `customer_region` derived from state code using macro-region mapping

```python
df_bronze = spark.table(f'bronze_customers')
df_bronze.show(30)
```

```python
df_bronze.printSchema()
#data types are appropriate
```

```python
nan_summary = df_bronze.select([
    spark_sum(col(c).isNull().cast("int")).alias(c)
    for c in df_bronze.columns
])
display(nan_summary)
#There are no null values in the customers table.
```

```python
display(df_bronze.groupBy("customer_id").count().filter(col("count") > 1))
#every customer_id is unique and will serve as table's primary key.
```

```python
ID_count = df_bronze.agg(
    count('customer_unique_id').alias('customer_unique_id_count'),
    count('customer_id').alias('customer_id_count'),
    countDistinct('customer_unique_id').alias('customer_unique_id_distinct_count'),
    countDistinct('customer_id').alias('customer_id_distinct_count')
)
display(ID_count)
#customer_unique_id is tied to an individual customer; e.g., a customer who makes more than one purchase in the data set will have the same unique_id assigned to them. However, a distinct customer_id is generated for each transaction, so there are no duplicate values.
```

```python
df_silver = (

    df_bronze
    .withColumn(
        "customer_city_cleaned",
        standardize_city_udf(
            F.col("customer_city")
        )
    )

    # Apply high-confidence municipality mappings
    .withColumn(
        "customer_city_cleaned",
        F.coalesce(
            mapping_expr[F.col("customer_city_cleaned")],
            F.col("customer_city_cleaned")
        )
    )

    # Final whitespace cleanup
    .withColumn(
        "customer_city_cleaned",
        F.trim(
            F.regexp_replace(
                F.col("customer_city_cleaned"),
                r"\s+",
                " "
            )
        )
    )
)
# Review cleaned output
display(
    df_silver
        .groupBy("customer_city_cleaned")
        .count()
        .orderBy(F.desc("count"))
)
```

```python
df_silver = df_silver.withColumn(
    "zip_code_valid",
    (col("customer_zip_code_prefix") >= 1000) & 
    (col("customer_zip_code_prefix") <= 99999)
)
invalid_zips = df_silver.filter(~col("zip_code_valid"))
print(f"Invalid zip codes: {invalid_zips.count()}")
#There are also no erroneous 00000 or 99999 values.
```

```python
df_silver = df_silver.withColumn(
    "state_valid",
    col("customer_state").isin(valid_states)
)
invalid_states = df_silver.filter(~col("state_valid"))
print(f"Invalid customer states: {invalid_states.count()}")
#all 27 listed states correspond to actual state abbreviations in Brazil.
```

```python
region_map_expr = create_map([lit(x) for x in chain(*region_mapping.items())])
df_silver = df_silver.withColumn("customer_region", region_map_expr[col("customer_state")])
```

```python
# Add audit columns
df_silver = df_silver.withColumn("silver_processing_timestamp", current_timestamp())
df_silver = df_silver.withColumn("bronze_source_count", lit(df_bronze.count()))

# Write raw data to the silver layer
df_silver.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("silver_customers")
```

### Geolocation <a name="geolocation"></a>

**Source**: `bronze_geolocation` → `silver_geolocation`  
**Primary key**: `geolocation_zip_code_prefix` (after deduplication)  
**Note**: The raw geolocation table has 1,000,163 rows on account of multiple coordinate pairs per ZIP code prefix. This table is deduplicated to one row per ZIP by averaging all valid coordinates within each prefix.

**Transformations applied**:
- Coordinates validated against Brazilian geographic boundaries  
  (lat: -33.75 to 5.27, lng: -73.99 to -28.85)
- Rows with invalid coordinates dropped before averaging
- Coordinates averaged per ZIP prefix to produce a single representative location per ZIP
- City names standardised using `standardize_city_name()` UDF
- State codes validated against all 27 official Brazilian state codes

```python
df_bronze = spark.table(f'bronze_geolocation')
df_bronze.show(30)
```

```python
df_bronze.printSchema()
```

```python
nan_summary = df_bronze.select([
    spark_sum(col(c).isNull().cast("int")).alias(c)
    for c in df_bronze.columns
])
display(nan_summary)
#no null values
```

```python
display(df_bronze.groupBy("geolocation_zip_code_prefix").count().filter(col("count") > 1))
#zip_code is the only field that can serve as a key for linking to other tables, but there are many duplicate values (as would be expected).
```

```python
# Brazilian coordinate range verification
df_silver = df_bronze.withColumn(
    "coordinates_valid",
    (col("geolocation_lat") >= -33.75) & 
    (col("geolocation_lat") <= 5.27) &
    (col("geolocation_lng") >= -73.99) &
    (col("geolocation_lng") <= -28.85)
)
invalid_coords = df_silver.filter(~col("coordinates_valid"))
print(f"Invalid coordinates: {invalid_coords.count()}")
```

```python
df_invalid_coords = df_silver.filter(~col("coordinates_valid"))
display(df_invalid_coords)
# none of the flagged coordinates are in Brazil. They will be dropped from the silver table.
```

```python
#drop invalid coordinates
df_silver = df_silver.filter(col("coordinates_valid"))
```

```python
df_silver = df_silver.withColumn(
    "state_valid",
    col("geolocation_state").isin(valid_states)
)
invalid_states = df_silver.filter(~col("state_valid"))
print(f"Invalid geolocation states: {invalid_states.count()}")
#all listed states correspond to actual state abbreviations in Brazil.
```

```python
df_silver = (

    df_silver
    .withColumn(
        "geolocation_city_cleaned",
        standardize_city_udf(
            F.col("geolocation_city")
        )
    )

    # Apply high-confidence municipality mappings
    .withColumn(
        "geolocation_city_cleaned",
        F.coalesce(
            mapping_expr[F.col("geolocation_city_cleaned")],
            F.col("geolocation_city_cleaned")
        )
    )

    # Final whitespace cleanup
    .withColumn(
        "geolocation_city_cleaned",
        F.trim(
            F.regexp_replace(
                F.col("geolocation_city_cleaned"),
                r"\s+",
                " "
            )
        )
    )
)
# Review cleaned output
display(
    df_silver
        .groupBy("geolocation_city_cleaned")
        .count()
        .orderBy(F.desc("count"))
)
```

```python
df_silver = df_silver.withColumn(
    "zip_code_valid",
    (col("geolocation_zip_code_prefix") >= 1000) & 
    (col("geolocation_zip_code_prefix") <= 99999)
)
invalid_zips = df_silver.filter(~col("zip_code_valid"))
print(f"Invalid zip codes: {invalid_zips.count()}")
#There are also no erroneous 00000 or 99999 values.
```

```python
df_zip_avg = df_silver.groupBy(
    "geolocation_zip_code_prefix"
).agg(
    avg("geolocation_lat").alias("geolocation_lat_updated"),
    avg("geolocation_lng").alias("geolocation_lng_updated"),
    count("*").alias("coordinate_samples")
)
```

```python
df_silver = (
    df_silver.join(
        df_zip_avg.select('geolocation_zip_code_prefix', 'geolocation_lat_updated', 'geolocation_lng_updated'),
        on='geolocation_zip_code_prefix',
        how='left'
    )
)
```

```python
# Drop unnecessary fields from df_silver
fields = [
    "geolocation_lat_updated",
    "geolocation_lng_updated",
    "geolocation_zip_code_prefix",
    "geolocation_city_cleaned",
    "geolocation_state",
    "ingestion_timestamp"
]

df_silver = df_silver.select(*fields)
```

```python
# Add processing metadata
df_silver = df_silver.withColumn("silver_processing_timestamp", current_timestamp())
df_silver = df_silver.withColumn("bronze_source_count", lit(df_bronze.count()))

# Write raw data to the silver layer
df_silver.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("silver_geolocation")
```

### Order Payment

**Source**: `bronze_order_payments` → `silver_order_payments`  
**Composite key**: `(order_id, payment_sequential)`: Multiple payment records per order are expected and valid (e.g. a voucher applied alongside a credit card payment).

**Transformations applied**:
- Payment types standardised to title case with human-readable labels (`credit_card` → `Credit Card`)
- 11 invalid payment rows removed or corrected:
  - Zero-value payments (likely failed vouchers or declined cards) removed
  - Positive-value payments with missing installment count defaulted to 1

```python
df_bronze = spark.table(f'bronze_order_payments')
df_bronze.show(30)
```

```python
df_bronze.printSchema()
```

```python
nan_summary = df_bronze.select([
    spark_sum(col(c).isNull().cast("int")).alias(c)
    for c in df_bronze.columns
])
display(nan_summary)
#no null values
```

```python
display(df_bronze.groupBy("order_id").count().filter(col("count") > 1))
#Once again, there are multiple instances of the same order_id. It can be used as a foreign key if other field values are aggregated.
```

```python
df_duplicates = df_bronze.groupBy("order_id").count().filter(col("count") > 1)
display(df_bronze.join(df_duplicates, "order_id"))
#For this table, the multiple instances of the identical order_id seems to be due to single purchases being split between vouchers and credit cards. Standard for sales data, and not an anomaly that requires cleaning.
```

```python
numeric_fields = ['payment_sequential', 'payment_installments', 'payment_value']

df_bronze.select(
    *[min(col).alias(f"{col}_min") for col in numeric_fields],
    *[max(col).alias(f"{col}_max") for col in numeric_fields]
).display()
#The min-max distribution is reasonable given what the categories quantify
```

```python
display(df_bronze.select('payment_type').distinct().orderBy('payment_type'))
#all payment types are mutually exclusive and can remain.
```

```python
#Edit category titles
df_silver = df_bronze.withColumn(
    'payment_type',
    when(col('payment_type') == 'boleto', 'Boleto')
    .when(col('payment_type') == 'credit_card', 'Credit Card')
    .when(col('payment_type') == 'debit_card', 'Debit Card')
    .when(col('payment_type') == 'voucher', 'Voucher')
    .otherwise('Not Defined')
)

display(df_silver.select('payment_type').distinct().orderBy('payment_type'))
```

```python
df_silver = df_silver.withColumn(
    "payment_valid",
    (col("payment_value") > 0) & (col("payment_installments") >= 1)
)
invalid_payments = df_silver.filter(~col("payment_valid"))
print(f"Invalid payments: {invalid_payments.count()}")
```

```python
display(df_silver.filter(~col("payment_valid")))
#The 9 valueless payments are attributable to voucher or not defined payment methods.
#The 2 payment amounts with recorded values were processed by credit card.
```

```python
orders_with_invalid = invalid_payments.select("order_id").distinct()
all_payments_for_these_orders = df_silver.join(
    orders_with_invalid, 
    "order_id"
).select("order_id", "payment_sequential", "payment_type", "payment_value", "payment_installments")

all_payments_for_these_orders.orderBy("order_id", "payment_sequential").show(50, truncate=False)
#Upon further inspection, most of the valueless payments were part of a larger order. They probably represent either valueless vouchers or card payments that were declined/did not process correctly. In any case, they do not provide any sales value to the table, and since order_payments concern processed sales, these rows can be removed from the table.
#The other entries are treated as data entry errors. Since payment went through, they will default to 1
```

```python
#Remove/clean the 11 flagged errors and default payment_installments to 1 where appropriate
df_silver = df_silver.filter(
    ~((col("payment_installments") == 1) & (col("payment_value") == 0))
).withColumn(
    "payment_installments",
    when(
        (col("payment_value") > 0) & 
        ((col("payment_installments") == 0) | col("payment_installments").isNull()),
        1  # Default to single payment
    ).otherwise(col("payment_installments"))
)
```

```python
# Add processing metadata
df_silver = df_silver.withColumn("silver_processing_timestamp", current_timestamp())
df_silver = df_silver.withColumn("bronze_source_count", lit(df_bronze.count()))

# Write raw data to the silver layer
df_silver.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("silver_order_payments")
```

### Order Reviews

**Source**: `bronze_order_reviews` → `silver_order_reviews`  
**Primary key**: `review_id`  
**Note on multiple reviews per order**: Olist allows customers to submit one review per item. Orders containing multiple items may therefore have multiple review records. This is expected behaviour. Review scores are later averaged at the order level in the gold layer.

**Transformations applied**:
- Rows with NULL or malformed `order_id` values removed (timestamp strings 
  entered in the wrong field were found in this column)
- `review_score` field found to contain mixed data: valid integer scores 
  (1–5) alongside timestamp strings entered in error. A cleaned integer 
  column `review_score_cleaned` was derived; rows with no valid score removed
- Original `review_score` column dropped in favour of `review_score_cleaned`

- Payment totals reconciled against order item totals per `order_id`, with a 1-cent tolerance for rounding differences
